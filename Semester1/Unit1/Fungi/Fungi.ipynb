# Question:
### What percent of mushrooms in this dataset are poisonous?

## Type of study:
### Census -- I will look at all the mushrooms in the dataset and determine which are poisonous and then determine the percentage of the poisinous mushrooms out of the whole dataset

```python
import pandas as pd
from IPython.display import Markdown, display
```

```python
fungi_data = pd.read_csv("mushrooms.csv")
```

```python
fungi_data.head()
```

```python
fungi_data.info()
```

```python
fungi_data["class"].describe()
```

```python
fungi_data["class"].value_counts()
```

```python
survived = fungi_data[fungi_data["class"] == "p"]
```

```python
survived.head()
```

```python
survived.info()
```

```python
poison_number = survived.shape[0]
```

```python
total_number = fungi_data.shape[0]
```

```python
rf_of_class = poison_number / total_number * 100
```

```python
display(Markdown("# Conclustion"))
display(Markdown(f"### The amount of poison mushrooms in this dataset is {round(rf_of_class, 3)}%"))
```
