# Question:
### What percent of mushrooms in this dataset are poisonous?

## Type of study:
### Census -- I will look at all the mushrooms in the dataset and determine which are poisonous and then determine the percentage of the poisinous mushrooms out of the whole dataset

```python
import pandas as pd
from IPython.display import Markdown, display
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
sns.set_theme()
```

```python
fungi_data = pd.read_csv("mushrooms.csv")
```

```python
# fungi_data.head()
```

```python
# fungi_data.info()
```

```python
# fungi_data["class"].describe()
```

```python
poison_number = fungi_data["class"].value_counts()["p"]
print(poison_number)
```

```python
total_number = fungi_data.shape[0]
```

```python
rf_of_class = poison_number / total_number * 100
```

```python
display(Markdown("# Conclustion"))
display(Markdown(f"### The relative frequency of poison mushrooms in this dataset is {round(rf_of_class, 3)}% (rounded to the 100ths place)"))
```

```python
sns.countplot(data=fungi_data, x="class")
```

```python
counts = fungi_data["class"].value_counts()
plt.figure(figsize=(6, 6))
plt.pie(
    counts,
    labels=counts.index.map({"e": "edible", "p": "poisonous"}),
    autopct="%1.1f%%",
    startangle=90,
    colors=["#66c2a5", "#fc8d62"]
)
plt.title("Mushroom Class Distribution")
plt.axis("equal")  # keeps the pie circular
plt.show()
```
