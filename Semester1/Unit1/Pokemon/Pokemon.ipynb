# Question: 
### How does Aerodactyl's speed stat compare to other pokemon?

## Study:
### This study is a cencus because it is looking at all pokemon

```python
import pandas as pd
import seaborn as sb
import matplotlib as mb
```

```python
pokemon = pd.read_csv("pokemon.csv")
```

```python
speed = pokemon["speed"]
aerodactyl = speed[141]
mean = speed.mean()
median = speed.median()
Q1 = speed.quantile(.25)
Q3 = speed.quantile(.75)
IQR = Q3 - Q1
standard_dev = speed.std()
print(standard_dev + mean + mean)
2 * standard_dev + mean
```

```python
plot = sb.histplot(
    data=speed,
)
plot.axvline(aerodactyl, color="purple", linestyle="--", linewidth=1, label=f"Aerodactyl: {aerodactyl:.2f}")
plot.axvline(mean, color="red", linestyle="--", linewidth=1, label=f"Mean: {mean:.2f}")
plot.axvline(median, color="yellow", linestyle="--", linewidth=1, label=f"Median: {median:.2f}")
plot.axvline(Q1, color="green", linestyle="--", linewidth=1, label=f"Q1: {Q1:.2f}")
plot.axvline(Q3, color="blue", linestyle="--", linewidth=1, label=f"Q3: {Q3:.2f}")
```
