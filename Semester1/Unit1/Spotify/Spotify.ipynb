# Question:
### What is the average tempo of these songs?

## Type of Study:
### Observational -- This dataset does not have every song released in it

```python
import pandas as pd
import seaborn as sb
```

```python
music_data = pd.read_csv("spotify-2023.csv", encoding="latin-1")
```

```python
# music_data.info()
```

```python
min = music_data["bpm"].min()
max = music_data["bpm"].max()
Q1 = music_data["bpm"].quantile(.25)
Q2 = music_data["bpm"].quantile(.75)
median = music_data["bpm"].median()
mean = music_data["bpm"].mean()
median2 = music_data["energy_%"].median()
mean2 = music_data["energy_%"].mean()
print(min)
print(max)
print(Q1)
print(Q2)
print(median)
print(round(mean))
```

```python
plot = sb.histplot(
    x=music_data["bpm"],
    y=music_data["energy_%"],
    binwidth=1,
    # kde=True
)

plot.axvline(mean, color="red", linestyle="--", linewidth=1, label=f"Mean: {mean:.2f}")
plot.axvline(median, color="green", linestyle="--", linewidth=1, label=f"Median: {median:.2f}")
plot.axhline(mean2, color="red", linestyle="--", linewidth=1, label=f"Mean2: {mean2:.2f}")
plot.axhline(median2, color="green", linestyle="--", linewidth=1, label=f"Median2: {median2:.2f}")

plot.set_xlabel("bpm")
```

# Conclusion: 
### There is very little correlation between the energy and tempo of the songs.
