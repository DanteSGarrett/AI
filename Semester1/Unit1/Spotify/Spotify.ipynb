# Question:
### What is the median release year?

## Type of Study:
### Observational -- This dataset does not have every song released in it

import pandas as pd
import seaborn as sb

```python
music_data = pd.read_csv("spotify-2023.csv", encoding="latin-1")
```

```python
# music_data.info()
```

```python
min = music_data["released_year"].min()
max = music_data["released_year"].max()
Q1 = music_data["released_year"].quantile(.25)
Q2 = music_data["released_year"].quantile(.75)
median = music_data["released_year"].median()
print(min)
print(max)
print(Q1)
print(Q2)
print(median)
```

```python
sb.histplot(
    data=music_data["released_year"],
    binwidth=.2
)
```

# Conclusion:
### Only 25% of the songs in this data were released before 2020. This dataset is heavily skewed to the left.
