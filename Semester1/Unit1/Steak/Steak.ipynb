```python
def check(array, index):
    if(array[index] == None):
        return ""
    else:
        return array[index]
```

```python
test1 = steak[steak.columns[n]].dropna().unique()
print(test1)
pyarrow.array()
```

```python
# Relative Frequency
unique_vals = steak[steak.columns[n]].dropna().unique()
if(steak.columns[n] != "ID Number"):
    # if(steak.columns[n] != ""): # for string bool vals
    #     var1 = unique_vals[0]
    #     var2 = unique_vals[1]
    # elif:
    #     var1=
    #     var2
    #     var3
    
        
    counts = steak[steak.columns[n]].value_counts()
    plot = mb.pyplot
    plot.figure(figsize=(6, 6))
    plot.pie(
        counts, 
        labels=counts.index.map({
            check(unique_vals, 0): check(unique_vals, 0), 
            check(unique_vals, 1): check(unique_vals, 1),
            check(unique_vals, 2): check(unique_vals, 2)}),
        autopct="%1.1f%%",
        startangle=90,
        colors=["#66c2a5", "#fc8d62"]
    )
    if(steak.columns[n] == "Gender"):
        plot.title(f"{steak.columns[n]}")
    else:
        plot.title(f"Percent of people that {steak.columns[n]}")
    plot.axis("equal")
    plot.show()
else:
    display(Markdown("Pick a different category, This cannot display the information in a useful way"))
```
