### Question:
Do men(in this study) gamble more than women?

### Type of Study:
Observational -- This study is not looking at every single male or female in the world

### My Analysis

```python
import pandas as pd
import seaborn as sbs
import matplotlib as mb
from IPython.display import Markdown, display
```

```python
steak = pd.read_csv("steak.csv", skiprows=[1])
```

```python
steak.columns = ["ID Number", # Not Used
                 "Choice of Lottery", # 2 Unique Values
                 "Smoke", # 2 Unique Values
                 "Drink Alcohol", # 2 Unique Values
                 "Gamble", # 2 Unique Values
                 "Have been Skydiving", # 2 Unique Values
                 "Drive above Speed Limit", # 2 Unique Values
                 "Have ever Cheated in a Relationship", # 2 Unique Values
                 "Eat Steak", # 2 Unique Values
                 "Preffered Steak Temp.", # 5 Unique Values
                 "Gender", # 2 Unique Values
                 "Age", # 4 Unique Values
                 "Household Income", # 5 Unique Values
                 "Education", # 5 Unique Values
                 "Location"] # 9 Unique Values
```

```python
# bar for f, pie for rf, print with md for mode
```

```python
n = 11 - 1 # Max 15 - 1 -- default value is 11 - 1
m = 5 - 1 # Max 15 - 1, should not == n -- default value is 5-1
```

```python
# Mode
if(steak.columns[n] != "ID Number"):
    mode = steak[steak.columns[n]].mode()
    display(Markdown(f"The mode of {steak.columns[n]} is {mode[0]}"))
else:
    display(Markdown("Pick a different category, This cannot display the information in a useful way"))
```

```python
# Frequency
if(steak.columns[n] != "ID Number"):
    xaxis = steak.columns[n]
    hue = steak.columns[m]
    sbs.countplot(data=steak, x=xaxis, hue=hue)
else:
    display(Markdown("Pick a different category, This cannot display the information in a useful way"))
```

```python
# Relative Frequency
# unique_vals = steak[steak.columns[n]].dropna().unique()
if(steak.columns[n] != "ID Number"):
    counts = steak[steak.columns[n]].value_counts()
    plot = mb.pyplot
    plot.figure(figsize=(6, 6))
    plot.pie(
        counts, 
        labels=counts.index,
        autopct="%1.1f%%",
        startangle=90,
        colors=["red", "yellow", "blue", "purple", "orange", "green", "cyan", "beige", "teal"]
    )
    if(steak.columns[n] == "Gender"):
        plot.title(f"{steak.columns[n]}")
    else:
        plot.title(f"Percent of people that {steak.columns[n]}")
    plot.axis("equal")
    plot.show()
else:
    display(Markdown("Pick a different category, This cannot display the information in a useful way"))
```

### My Conclusion
More men said they gambled, while more women said they did not gamble.
