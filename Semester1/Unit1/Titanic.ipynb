```python
import pandas as pd
```

```python
# create the datatable
titanic = pd.read_csv("data/titanic.csv")
titanic
```

```python
titanic.head()
```

```python
# dataframe info
titanic.info()
```

```python
# One collumn
ages = titanic["Age"] # returns a pandas Series
ages.head() # first n entries (default 5)
```

```python
# multipe collumns
age_sex = titanic[["Age", "Sex"]] # returns a dataframe
age_sex.tail() # last 5 entries
```

```python
# specific rows with conditionals
above_35 = titanic[titanic["Age"] > 35]
above_35.head()
```

```python
# specific rows with .isin()
class_23 = titanic[titanic["Pclass"].isin([2, 3])]
class_23.head()
```

```python
# non-null values in the rows
age_no_na = titanic[titanic["Age"].notna()]
age_no_na.head()
```

```python
# specfic rows & collums
adult_names = titanic.loc[titanic["Age"] > 35, "Name"]
adult_names.head()
```

```python
# specific rows & columns with index numbers
titanic.iloc[9:25, 2:5] # rows 9-25 and collumns 2-5
titanic.head()
```

```python
survived = (titanic[titanic["Survived"] == 1])
count = len(survived)
print("Survived:", count)
```
