```python
import numpy as np
import pandas as pd


def add_relative_frequency(df, column_name, new_col_name="relative_frequency"):
    """Adds a relative frequency column for values in a specified dataframe column."""
    # Calculate value counts normalized to get relative frequencies (proportions)
    freq_map = df[column_name].value_counts(normalize=True)

    # Map the relative frequencies back to the original column values
    df[new_col_name] = df[column_name].map(freq_map)

    return df


# Example usage:
# Create dummy data with 1,000 rows ranging from 1 to 6
np.random.seed(42)
df = pd.DataFrame({"dice_rolls": np.random.randint(1, 7, size=1000)})

# Apply function
df = add_relative_frequency(df, "dice_rolls")

# Display the first few rows
print(df.head())
```

```python
import numpy as np
import pandas as pd


def add_cumulative_relative_frequency_of_one(
    df, column_name, new_col_name="cum_rel_freq_of_1"
):
    """Adds a running (cumulative) relative frequency column specifically for the value 1."""
    # Create a boolean series indicating where the value is 1
    is_one = df[column_name] == 1

    # Calculate cumulative sum of 1s divided by the cumulative row count (1-indexed)
    cumulative_count_of_ones = is_one.cumsum()
    total_rows_so_far = np.arange(1, len(df) + 1)

    df[new_col_name] = cumulative_count_of_ones / total_rows_so_far

    return df


# Example usage:
np.random.seed(3)
df = pd.DataFrame({"dice_rolls": np.random.randint(1, 7, size=1000)})

# Apply function
df = add_cumulative_relative_frequency_of_one(df, "dice_rolls")

# Display the first few rows
print(df.head(20))
```
